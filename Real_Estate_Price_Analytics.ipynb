# **Missing Data Treatment**

```python
df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].mean())
df['bathrooms'] = df['bathrooms'].fillna(df['bathrooms'].mean())
```

# **Step 3: Exploratory Data Analysis (EDA)**

Q1.What is the distrubution of house prices?

```python
plt.figure()
plt.hist(df['price'], bins= 30)
plt.title('Distribution of House Prices')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

# **Analysis :-**

- Distrubution is Right Skewed.

- Majority houses are in tha mid price range
# **Conclusion.**
The house-price distribution is highly right-skewed, with most properties concentrated in the lower-to-middle price ranges and relatively few properties at very high prices.

# **Q2. How does living area affect price?**

```python
plt.scatter(df['sqft_living'],df['price'])
plt.xlabel('Living Area (sqft)')
plt.title('Living Area vs. Price')
plt.show()
```

# **Analysis**
- a positive relationship between living area and house price.

# **Conclusion**
- Living area is an important factor influencing house prices.

# **Q3. What is the relationship between bedrooms and Price.**

```python
plt.figure()
sns.boxplot(x='bedrooms', y='price', data=df)
plt.xlabel('Bedrooms')
plt.ylabel('Price')
plt.title('Bedrooms vs. Price')
plt.show()
```

# **Analysis:**
- House prices generally tend to increase with the number of bedrooms, although the relationship is not consistent.

## **Conclusion:**
- Larger bedroom counts do not always guarantee higher prices.

# **Q4. Do bathrooms impack house price?**

```python
sns.scatterplot(x='bathrooms', y='price', data=df)
plt.xlabel('Bathrooms')
plt.ylabel('Price')
plt.title('Bathrooms vs. Price')
plt.show()
```

# **Analysis:**
- Property prices tend to increase as the number of bathrooms increases, but the relationship is not consistent.
# **conclusion:**

- There is a general positive relationship between bathrooms and property prices
- Bathrooms are more important than bedrooms for pricing.

# **Q5.Does waterfront view increase price?**

```python
sns.boxplot(x='waterfront', y='price', data=df)
plt.title('Waterfront vs. Price')
plt.show()
```

# **Analysis**
- Waterfront houses are much costlier

# **conclusion**
 - Waterfront properties tend to be more expensive than non-waterfront properties.

# **Q6.What is the correlation among key variables?**

```python
corr = df[['price','sqft_living','bathrooms','bedrooms','grade']].corr()
sns.heatmap(corr, annot=True)
plt.title('Correlation Matrix')
plt.show()
```

# **Analysis:**
- Square footage (0.70) has the strongest correlation with price, indicating that larger homes tend to be more expensive.
# **Conclusion:**
- Square footage is the strongest factor correlated with house prices among the variables shown, followed by grade and bathrooms.
